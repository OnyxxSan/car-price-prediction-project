# Car Price Prediction Analysis

This notebook explores a synthetic used-car pricing dataset and compares several regression models.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from src.car_price_pipeline import load_dataset

data = load_dataset()
data.head()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
sns.boxplot(data=data, x='fuel_type', y='price', palette='Set2', ax=ax)
plt.title('Car Price by Fuel Type')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
